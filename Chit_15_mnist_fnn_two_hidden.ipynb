```python
# Chit 15: Handwritten MNIST FNN: two hidden layers
# Step 1: Import the libraries
import numpy as np
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense
```

```python
# Step 2: Load and prepare Handwritten MNIST
from tensorflow.keras.datasets import mnist

(X_train, y_train), (X_test, y_test) = mnist.load_data()
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0
class_names = [str(i) for i in range(10)]
image_shape = (28, 28, 1)
```

```python
# Step 3: Flatten the images
X_train = X_train.reshape(len(X_train), 784)
X_test = X_test.reshape(len(X_test), 784)
```

```python
# Step 4: Build the model
num_classes = len(class_names)
model = Sequential([
    Input(shape=(X_train.shape[1],)),
    Dense(256, activation="relu"),
    Dense(128, activation="relu"),
    Dense(num_classes, activation="softmax")
])
```

```python
# Step 5: Compile and train
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
# Step 6: Prepare and run the model
model.fit(X_train, y_train, epochs=15, batch_size=128, verbose=0)
```

```python
# Step 7: Evaluate on test data
_, test_accuracy = model.evaluate(X_test, y_test, verbose=0)
print(f"Testing accuracy: {test_accuracy * 100:.2f}%")
```
