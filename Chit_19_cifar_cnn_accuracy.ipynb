```python
# Chit 19: CIFAR-10 CNN: object classification
# Step 1: Import the libraries
import numpy as np
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense
import pandas as pd
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten
from sklearn.model_selection import train_test_split
```

```python
# Step 2: Load and prepare CIFAR-10
data = pd.read_csv("cifar-10.csv", dtype=np.uint8)
y = data["label"].to_numpy(dtype="int64")
X = data.drop(columns="label").to_numpy(dtype="float32")
X = X.reshape(-1, 32, 32, 3) / 255.0

class_names = ["airplane", "automobile", "bird", "cat", "deer", "dog", "frog", "horse", "ship", "truck"]
image_shape = (32, 32, 3)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y
)
```

```python
# Step 3: Build the model
num_classes = len(class_names)
model = Sequential([
    Input(shape=X_train.shape[1:]),
    Conv2D(32, (3, 3), activation="relu"),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation="relu"),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation="relu"),
    Dense(num_classes, activation="softmax")
])
```

```python
# Step 4: Compile and train
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
# Step 5: Prepare and run the model
model.fit(X_train, y_train, epochs=10, batch_size=128, verbose=0)
```

```python
# Step 6: Evaluate on test data
_, test_accuracy = model.evaluate(X_test, y_test, verbose=0)
print(f"Testing accuracy: {test_accuracy * 100:.2f}%")
```
